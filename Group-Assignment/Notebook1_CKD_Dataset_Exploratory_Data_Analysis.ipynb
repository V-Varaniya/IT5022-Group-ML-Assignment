# Notebook 1: CKD Dataset Exploratory Data Analysis




## 1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Load the dataset

```python
df = pd.read_csv("ckd-dataset-v2.csv")

print("Original dataset shape:", df.shape)
display(df.head())
```

```python
display(df.info())
```

## 3.Inspect the dataset structure

```python
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nFirst 10 rows:")
display(df.head(10))

print("\nLast 5 rows:")
display(df.tail())
```

## 4. Inspect Target Column

```python
# Check target column values and their frequencies
print(df["class"].value_counts())

# Check missing values
print("Missing values:", df["class"].isnull().sum())

# Check unique target labels
print("Unique labels:", df["class"].unique())
```

```python
# Display rows with invalid or missing target values
invalid_rows = df[
    df["class"].isna() |
    ~df["class"].isin(["ckd", "notckd"])
]

display(invalid_rows)
```

```python
# Keep only records with valid target labels
df = df[df["class"].isin(["ckd", "notckd"])].copy()

# Verify the result
print(df["class"].value_counts())
print("Missing target values:", df["class"].isna().sum())
print("Unique target labels:", df["class"].unique())
print("Dataset shape:", df.shape)
```

## 5. Analyze the target class distribution

```python
class_counts = df["class"].value_counts()

print(class_counts)
print("\nClass percentages:")
print((df["class"].value_counts(normalize=True) * 100).round(2))

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="class", hue="class", legend=False)
plt.title("Distribution of CKD and Non-CKD Records")
plt.xlabel("Class")
plt.ylabel("Number of records")
plt.tight_layout()
plt.show()
```

## 6.Examine missing and unusual values in all columns

```python
# Count actual missing values
print("Missing values per column:")
display(df.isnull().sum().sort_values(ascending=False))

# Count empty strings and common missing markers
missing_markers = ["", "?", "NA", "N/A", "None", "null"]

for col in df.columns:
    values = df[col].astype("string").str.strip()
    count = values.isin(missing_markers).sum()
    if count > 0:
        print(f"{col}: {count} possible missing markers")

# Count distinct values per column
print("\nNumber of distinct values:")
display(df.nunique(dropna=False).sort_values())
```

## 7.Visualize missing values

```python
missing_pct = (
    df.isnull().mean().mul(100)
    .sort_values(ascending=False)
)

missing_pct = missing_pct[missing_pct > 0]

if not missing_pct.empty:
    plt.figure(figsize=(10, 5))
    missing_pct.plot(kind="bar")
    plt.title("Percentage of Missing Values by Feature")
    plt.ylabel("Missing values (%)")
    plt.xlabel("Feature")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
else:
    print("No actual NaN values found in the filtered data.")
```

## 8. Investigate feature categories and range values

```python
for col in df.columns:
    if col != "class":
        print(f"\n--- {col} ---")
        print(df[col].value_counts(dropna=False).head(15))
```

## 9.Explore categorical feature distributions

```python
features_to_plot = ["sg", "al", "bgr", "hemo", "grf", "age"]

for col in features_to_plot:
    if col in df.columns:
        plt.figure(figsize=(10, 4))

        order = df[col].value_counts().index

        sns.countplot(
            data=df,
            x=col,
            hue=col,
            order=order,
            legend=False
        )

        plt.title(f"Distribution of {col}")
        plt.xlabel(col)
        plt.ylabel("Number of records")
        plt.xticks(rotation=60, ha="right")
        plt.tight_layout()
        plt.show()
```

## 10.Compare feature categories across CKD and non-CKD

```python
features_to_compare = ["sg", "al", "bgr", "hemo", "grf", "age"]

for col in features_to_compare:
    if col in df.columns:
        proportions = pd.crosstab(
            df[col],
            df["class"],
            normalize="columns"
        ) * 100

        proportions.plot(
            kind="bar",
            figsize=(10, 4)
        )

        plt.title(f"{col} Categories by Class")
        plt.xlabel(col)
        plt.ylabel("Percentage within class")
        plt.xticks(rotation=60, ha="right")
        plt.tight_layout()
        plt.show()
```
